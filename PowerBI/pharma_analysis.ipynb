```python
# Side Effects Count
df_clean['Side_Effects_Count'] = df_clean['Side_Effects'].apply(
    lambda x: len(str(x).split(',')) if x != '' else 0
)

# Risk Score
df_clean['Risk_Score'] = df_clean['Side_Effects_Count']

# Complexity
df_clean['Complexity'] = df_clean['Ingredient_Count']

# Cost Efficiency
df_clean['Cost_Efficiency'] = 1 / df_clean['Final_Price']

# Risk-Cost Index
df_clean['Risk_Cost_Index'] = df_clean['Risk_Score'] * df_clean['Final_Price']

# Price Difference
df_clean['Price_Difference'] = df_clean['Price'] - df_clean['Final_Price']
```

```python
df_clean.head()
```

```python
risk_threshold = df_clean['Risk_Score'].mean()
price_threshold = df_clean['Final_Price'].mean()

risk_threshold, price_threshold
```

```python
danger_zone = df_clean[
    (df_clean['Risk_Score'] > risk_threshold) &
    (df_clean['Final_Price'] > price_threshold)
]
```

```python
danger_zone.sort_values(
    by='Risk_Cost_Index', ascending=False
)[['Medicine', 'Disease', 'Final_Price', 'Risk_Score']].head(10)
```

```python
disease_cost = df_clean.groupby('Disease').agg({
    'Final_Price': 'mean',
    'Medicine': 'count',
    'Risk_Score': 'mean'
}).reset_index()
```

```python
disease_cost.columns = [
    'Disease', 'Avg_Treatment_Cost', 'Medicine_Count', 'Avg_Risk'
]
```

```python
disease_cost['Disease_Burden'] = (
    disease_cost['Avg_Treatment_Cost'] * disease_cost['Medicine_Count']
)
```

```python
disease_cost.sort_values(
    by='Disease_Burden', ascending=False
).head(10)
```

```python
import matplotlib.pyplot as plt

top_diseases = disease_cost.sort_values(
    by='Disease_Burden', ascending=False
).head(10)

plt.figure(figsize=(10,6))
plt.barh(top_diseases['Disease'], top_diseases['Disease_Burden'])

plt.xlabel("Disease Burden")
plt.ylabel("Disease")
plt.title("Top 10 Most Expensive Diseases")

plt.gca().invert_yaxis()

plt.show()
```

- Business Question: Which diseases have the highest overall treatment cost burden?

- Insight:
  - Cancer-related treatments have significantly higher cost compared to other diseases.
  - Disease burden is driven by both high medicine prices and number of prescriptions.

- Impact:
  Helps identify high-cost diseases where cost optimization strategies are most needed. **bold text**

```python
from sklearn.preprocessing import StandardScaler

features = df_clean[['Final_Price', 'Risk_Score', 'Ingredient_Count']]
```

```python
scaler = StandardScaler()
scaled_features = scaler.fit_transform(features)
```

```python
from sklearn.cluster import KMeans

kmeans = KMeans(n_clusters=3, random_state=42)
df_clean['Cluster'] = kmeans.fit_predict(scaled_features)
```

```python
df_clean[['Medicine', 'Final_Price', 'Risk_Score', 'Ingredient_Count', 'Cluster']].head(10)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,6))
plt.scatter(df_clean['Final_Price'], df_clean['Risk_Score'], c=df_clean['Cluster'])

plt.xlabel("Final Price")
plt.ylabel("Risk Score")
plt.title("Medicine Clusters")

plt.show()
```

- Business Question: Can medicines be grouped based on price and risk?

- Insight:
  - Medicines can be divided into groups like low-risk, high-risk, and high-cost.
  - Some medicines are both expensive and risky, which may not be the best choice.

- Impact:
  Helps in choosing safer and cheaper medicines for better treatment. **bold text**

```python
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

X = df_clean[['Final_Price', 'Ingredient_Count']]
y = df_clean['Risk_Score']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)

model = RandomForestClassifier()
model.fit(X_train, y_train)

model.score(X_test, y_test)
```

```python
import pandas as pd

importance = pd.Series(model.feature_importances_, index=X.columns)
importance.sort_values(ascending=False)
```
